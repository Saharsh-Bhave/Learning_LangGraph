```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langgraph.checkpoint.memory import InMemorySaver
```

```python
load_dotenv()

llm = ChatOpenAI()
```

```python
class JokeState(TypedDict):

    topic: str
    joke: str
    explanation: str
```

```python
def gen_joke(state: JokeState):

    prompt = f'Generate a joke on the topic {state["topic"]}.'

    response = llm.invoke(prompt).content

    return {'joke': response}
```

```python
def gen_expln(state: JokeState):

    prompt = f'Write an explanation for the joke - {state["joke"]}'
    response = llm.invoke(prompt).content

    return {'explanation': response}
```

```python
graph = StateGraph(JokeState)

graph.add_node('gen_joke', gen_joke)
graph.add_node('gen_expln', gen_expln)

graph.add_edge(START, 'gen_joke')
graph.add_edge('gen_joke', 'gen_expln')
graph.add_edge('gen_expln', END)

checkpointer = InMemorySaver()

workflow = graph.compile(checkpointer=checkpointer)
```

```python
config1 = {'configurable': {'thread_id': '1'}}
workflow.invoke({'topic': 'soccer'}, config=config1)
```

```python
#Final state value

workflow.get_state(config1)
```

```python
#intermediate state values

list(workflow.get_state_history(config1))
```
